## **Part 3: The pipeline with other libraries**

While spaCy is the most convenient solution for running an entire NLP pipeline, [NLTK](https://www.nltk.org/) also offers support for the individual steps.

```
from nltk.tokenize import word_tokenize
from nltk.tag.perceptron import PerceptronTagger
from nltk.stem import WordNetLemmatizer

#Tokenization
tokenized = word_tokenize(example_sentence)

#POS Tagging
#Initialize POS Tagger
tagger = PerceptronTagger()

#Perform taggging
pos_tags = tagger.tag(word_tokenize(example_sentence))
print(f"POS Tags: {pos_tags}\n")

# Initialize lemmatizer
lemmatizer = WordNetLemmatizer()

#Perform lemmatization
print("Lemmatized:")
for token in tokenized:
  lemmatized = lemmatizer.lemmatize(token)
  print(lemmatized)
```

NLTK does not only offers lemmatization but also different versions of stemming.

```
from nltk.stem.porter import PorterStemmer
from nltk.stem.lancaster import LancasterStemmer
from nltk.stem import SnowballStemmer

# Initializing stemmers
porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")

example_word = "organization"

print(porter.stem(example_word))
print(lancaster.stem(example_word))
print(snowball.stem(example_word))
```

The same can be accomplished with other lirbaries, such as [Stanza from Stanford](http://stanza.run/).

While there are some options for dependency parsing in NLTK, the successful ones largely depend on the StanfordNLP Parser. Thus, we will directly use StanfordNLP and run all the previous steps in the socalled pipeline.

First, we need to download the English model for StanfordNLP.

```
!pip install stanza
```

```
import stanza
stanza.download("en")
```

Then we can run the library on the processing steps we are interested in.

❓ Is there any difference in the pipeline outputs between spaCy and Stanza?



```
pipeline = stanza.Pipeline(lang='en', processor='tokenize,pos,lemma,depparse')

doc = pipeline(example_sentence)

for sentence in doc.sentences:
  print(sentence.print_dependencies())
  for word in sentence.words:
    print(f"Text: {word.text} | POS: {word.upos} | Lemma: {word.lemma}")
```
