### Question 4: What does the following code snippet print?

```python
text = "the cat sat on the mat"
words = text.split()

counts = {}
for w in words:
    #counts[w] = counts.get(w, 0) + 1
    if w not in counts.keys():
      counts[w] = 1
    else:
      counts[w] += 1
print(counts["the"])
```
