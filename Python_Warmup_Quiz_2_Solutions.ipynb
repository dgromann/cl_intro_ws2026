### Question 1: What does the following code snippet print?







```python
def tokenize(text, lowercase=True):
    if lowercase:
        text = text.lower()
    return text.split()

print(tokenize("The Cat sat"))
print(tokenize("The Cat sat", lowercase=False))
```

### Question 2: What needs to be fixed in the following code snippet?

```python
#Solution 1: Replace + by comma
def add_one(n):
    return n + 1

#print("Result: " + add_one(4))
print("Result: ", add_one(4))
```

```python
#Solution 2: Convert result to string so that + can be used
def add_one(n):
    return n + 1

#print("Result: " + add_one(4))
print("Result: " + str(add_one(4)))
```

```python
#Solution 3: Use f string instead of plus in the print statement
def add_one(n):
    return n + 1

#print("Result: " + add_one(4))
print(f"Result: {str(add_one(4))}")
```

### Question 5: What needs to be fixed in the following code snippet?

```python
# Solution 1: Add a default value to the function
def greet(name, city="Innsbruck"):
    print("Hi", name, "from", city)

greet("Anna")
```

```python
# Solution 1: Add a city value to the line calling the greet function
def greet(name, city):
    print("Hi", name, "from", city)

greet("Anna", "Innsbruck")
```

### Question 6: What does the code cell print now?

```python
def describe(word, n=3):
    if len(word) > n:
        return word + " is long"
    return word + " is short"

print(describe("cat"))
print(describe("cat", 2))
print(describe("language"))
```
