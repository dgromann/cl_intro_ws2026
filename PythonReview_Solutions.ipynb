## Dictionary

Dictionaries are hash maps. They contain key-value pairs, which both can be addressed separately. In the following code cell, we create an empty dictionary.

```
grades = {}
grades = dict()
```

We can now add the first and the second item to the empty dictionary, using two different methods of how to do that. An item is always a key-value pair.

```
grades = {"Elon" : "5"}
grades["Mary"] = "1"
print(grades)
```

To check if a specific key is in the dictionary already, we can use the following code:

```
print("Elon" in grades)
```

If you wish to get the value that is associated with a specific key, you can use the following code. This is similar to the lookup of information associated with a natural language word in an actual monolingual or bilingual dictionary.

❓ Question 6: What will the result be of the following code cell?

```
print(grades["Elon"])
```

This is how you can change the value associated with a specific key:

```
grades["Elon"] = 4
print(grades)
```

If you wish to iterate over all key-value pairs in a dictionary, you can use the function `items()` in a `for` loop.

```
for key, value in grades.items():
  print(key, value)
```

If you only wish to iterate over and process all the values or keys of the dictionary, you can use the following commands:

```
for value in grades.values():
  print(value)

for key in grades.keys():
  print(key)
```

## **Part 3: Functions**

A function is a piece of code with a name that allows you to reuse that code where you need it.

We can count the number of words in a sentence as follows, if we only want to do it once.

```
sentence = "the cat sat on the mat"
words = sentence.split()
print(len(words))
```

But if we wish to repeat this process and count the number of words in many different sentences, we might want to write a function for it. The function needs to be defined with `def`followed by a name you assign and brackets `()`.

The value in the bracket is called the argument and handed over for the function when you call it. Here we give the function a sentence to count the number of words in it. A return function sends the response after processing the input back to the code where you call the function.

With this function, we can reuse the `count_words` snippet of code for several sentences, which is even more useful for longer snippets.

```
def count_words(sentence):
  words = sentence.split()
  return len(words)


sentences_list = ["The cat sat on the mat",
                  "The cat was in front of the fire",
                  "The house is small"]

for sentence in sentences_list:
  print(count_words(sentence))
```

Once a function is defined in a Python runtime environment, you can reuse it across code cells in Colab just like other variables:

```
print(count_words("This is another example sentence"))
```

👋 ⚒ Exercise 13: Write a word frequency function that takes a text as input and returns a dictionary with the words as keys and the frequency count as values.

❗Keep in mind that words should be lowercased for an accurate frequency count.

```
def word_frequencies(text):
  frequency_count = {}
  for word in text.lower().split():
    if word in frequency_count.keys():
      frequency_count[word] += 1
    else:
      frequency_count[word] = 1
    #Alternative one-line code for the if-else statement:
    #frequency_count[word] = frequency_count.get(word, 0) + 1
  return frequency_count

text = """Row, row, row your boat,
Gently down the stream.
Merrily, merrily, merrily, merrily,
Life is but a dream. """

print(word_frequencies(text))
```
